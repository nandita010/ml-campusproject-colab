## Weekly Decision & Justification Log
Document at least one significant technical or project decision each week. Exceptional weeks may contain more than one decision, but quality is valued over quantity.

Keep the decision log as one section in your report, adding to it weekly. The main things to consider are:
- **Issue or challenge identified**; Describe the most important issue or challenge identified this week
- **Decision taken**; State the decision or approach taken
- **Justification**; Explain why this approach was selected. What other approaches were considered?
- **Evidence supporting the decision**; What evidence was used to justify the decision?
- **AI Use**; What AI too was used (if any)? Describe how it assisted. What suggestion(s) were adopted or not adopted?
- **Reflection**; What/So what/Now what?

The table below shows one way you could format your log - but you don't have to stick to this format. Guideline: one/two sentences for each entry.
